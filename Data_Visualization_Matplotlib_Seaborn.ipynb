# Data Visualization with Matplotlib & Seaborn

**Project objective:** Create informative visualizations from a real-world dataset using Python, Pandas, Matplotlib, and Seaborn.

We will use the Titanic passenger dataset and create:
- Histograms and distribution plots
- Count plots and bar charts
- Box plots
- Scatter plots
- Correlation matrix and heatmap
- Pair plot

**Instructions:** Run each cell from top to bottom.

## 1. Import required libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid", palette="deep")

print("Libraries imported successfully!")
```

## 2. Load a real dataset

```python
# Load Titanic dataset from Seaborn
df = sns.load_dataset("titanic")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
display(df.head())
```

## 3. Understand the dataset

```python
print("Dataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe(include="all").T)

print("\nMissing values:")
display(df.isnull().sum().to_frame("Missing values"))
```

## 4. Prepare data for visualization

```python
# Work on a copy so the original dataset remains unchanged
data = df.copy()

# Fill missing numeric age values with the median
data["age"] = data["age"].fillna(data["age"].median())

# Fill missing categorical values with the mode
for col in ["embarked", "embark_town"]:
    data[col] = data[col].fillna(data[col].mode()[0])

# Create readable labels for survival
data["survival_status"] = data["survived"].map({
    0: "Did not survive",
    1: "Survived"
})

print("Data prepared for visualization.")
display(data[["age", "sex", "pclass", "fare", "survival_status"]].head())
```

## 5. Histogram – Passenger age distribution

```python
plt.figure(figsize=(9, 5))
sns.histplot(data=data, x="age", bins=30, kde=True)
plt.title("Distribution of Passenger Ages")
plt.xlabel("Age (years)")
plt.ylabel("Number of passengers")
plt.tight_layout()
plt.show()
```

## 6. Histogram – Ticket fare distribution

```python
plt.figure(figsize=(9, 5))
sns.histplot(data=data, x="fare", bins=35, kde=True)
plt.title("Distribution of Ticket Fares")
plt.xlabel("Fare")
plt.ylabel("Number of passengers")
plt.tight_layout()
plt.show()
```

## 7. Count plot – Passenger survival

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=data, x="survival_status")
plt.title("Number of Passengers by Survival Status")
plt.xlabel("Survival status")
plt.ylabel("Passenger count")
plt.tight_layout()
plt.show()
```

## 8. Bar chart – Survival rate by gender

```python
gender_rate = data.groupby("sex", observed=False)["survived"].mean().mul(100).reset_index()

plt.figure(figsize=(7, 5))
sns.barplot(data=gender_rate, x="sex", y="survived")
plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival rate (%)")
plt.ylim(0, 100)
plt.tight_layout()
plt.show()

display(gender_rate.rename(columns={"survived": "survival_rate_percent"}))
```

## 9. Box plot – Age by passenger class

```python
plt.figure(figsize=(9, 5))
sns.boxplot(data=data, x="pclass", y="age")
plt.title("Age Distribution Across Passenger Classes")
plt.xlabel("Passenger class")
plt.ylabel("Age (years)")
plt.tight_layout()
plt.show()
```

## 10. Box plot – Fare by survival status

```python
plt.figure(figsize=(9, 5))
sns.boxplot(data=data, x="survival_status", y="fare")
plt.title("Ticket Fare by Survival Status")
plt.xlabel("Survival status")
plt.ylabel("Ticket fare")
plt.tight_layout()
plt.show()
```

## 11. Scatter plot – Age vs fare

```python
plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=data,
    x="age",
    y="fare",
    hue="survival_status",
    alpha=0.7
)
plt.title("Passenger Age vs Ticket Fare")
plt.xlabel("Age (years)")
plt.ylabel("Fare")
plt.tight_layout()
plt.show()
```

## 12. Correlation matrix

```python
# Select numeric columns for correlation analysis
numeric_data = data.select_dtypes(include=np.number)
correlation_matrix = numeric_data.corr()

display(correlation_matrix.round(2))
```

## 13. Correlation heatmap

```python
plt.figure(figsize=(11, 7))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    center=0
)
plt.title("Correlation Heatmap of Titanic Features")
plt.tight_layout()
plt.show()
```

## 14. Pair plot – Relationships between numeric features

```python
# Use a sample to keep the pair plot fast and readable
pair_data = data[["age", "fare", "sibsp", "parch", "survival_status"]].dropna()

sns.pairplot(
    pair_data.sample(min(250, len(pair_data)), random_state=42),
    hue="survival_status",
    diag_kind="hist"
)
plt.suptitle("Pair Plot of Selected Titanic Features", y=1.02)
plt.show()
```

## 15. Heatmap – Survival rate by gender and class

```python
survival_pivot = data.pivot_table(
    index="sex",
    columns="pclass",
    values="survived",
    aggfunc="mean",
    observed=False
) * 100

plt.figure(figsize=(8, 5))
sns.heatmap(
    survival_pivot,
    annot=True,
    cmap="YlGnBu",
    fmt=".1f",
    cbar_kws={"label": "Survival rate (%)"}
)
plt.title("Survival Rate by Gender and Passenger Class")
plt.xlabel("Passenger class")
plt.ylabel("Gender")
plt.tight_layout()
plt.show()
```

## 16. Save visualizations

```python
# Save selected charts as image files
plt.figure(figsize=(9, 5))
sns.histplot(data=data, x="age", bins=30, kde=True)
plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Passenger count")
plt.tight_layout()
plt.savefig("titanic_age_histogram.png", dpi=300, bbox_inches="tight")
plt.show()

print("Chart saved as titanic_age_histogram.png")
```

## 17. Key statistical insights

```python
print("========== DATA VISUALIZATION SUMMARY ==========")
print("Total passenger records:", len(data))
print("Average passenger age:", round(data["age"].mean(), 2))
print("Median passenger age:", round(data["age"].median(), 2))
print("Average ticket fare:", round(data["fare"].mean(), 2))
print("Overall survival rate:", round(data["survived"].mean() * 100, 2), "%")

print("\nSurvival rate by gender (%):")
print((data.groupby("sex", observed=False)["survived"].mean() * 100).round(2))

print("\nSurvival rate by passenger class (%):")
print((data.groupby("pclass")["survived"].mean() * 100).round(2))
print("================================================")
```
