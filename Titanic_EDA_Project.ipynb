# Exploratory Data Analysis (EDA) on Titanic

**Objective:** Explore the Titanic passenger dataset using Pandas, NumPy, Matplotlib, and Seaborn. Clean missing values, calculate statistics, visualize patterns, and summarize findings.

Run the cells from top to bottom.

## 1. Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

## 2. Load the Titanic dataset

```python
# Load the classic Titanic dataset provided by Seaborn
df = sns.load_dataset("titanic")

print("Dataset loaded successfully!")
display(df.head())
```

## 3. Explore the dataset

```python
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe(include="all").T)
```

## 4. Check missing values

```python
missing = df.isnull().sum().sort_values(ascending=False)
display(missing[missing > 0].to_frame("Missing values"))

plt.figure(figsize=(12, 5))
sns.heatmap(df.isnull(), cbar=False, cmap="viridis")
plt.title("Missing Values in Titanic Dataset")
plt.xlabel("Columns")
plt.ylabel("Passenger records")
plt.show()
```

## 5. Clean the data

```python
# Fill numerical age values with the median
df["age"] = df["age"].fillna(df["age"].median())

# Fill categorical missing values with the most frequent value
for column in ["embarked", "embark_town"]:
    df[column] = df[column].fillna(df[column].mode()[0])

# Drop deck because it contains many missing values
df = df.drop(columns=["deck"])

# Check remaining missing values
print("Missing values after cleaning:")
display(df.isnull().sum().to_frame("Missing values"))

print("Shape after cleaning:", df.shape)
```

## 6. Overall survival analysis

```python
survival_counts = df["survived"].value_counts().sort_index()
survival_percentages = df["survived"].value_counts(normalize=True).sort_index() * 100

print("Survival counts (0 = did not survive, 1 = survived):")
display(survival_counts.to_frame("Passengers"))

print("Survival percentages:")
display(survival_percentages.round(2).to_frame("Percentage"))

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="survived")
plt.title("Titanic Survival Distribution")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of passengers")
plt.show()
```

## 7. Survival by gender

```python
gender_survival = df.groupby("sex", observed=False)["survived"].mean() * 100
print("Survival rate by gender (%):")
display(gender_survival.round(2).to_frame("Survival rate (%)"))

plt.figure(figsize=(7, 5))
sns.barplot(data=df, x="sex", y="survived", errorbar=None)
plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival rate")
plt.ylim(0, 1)
plt.show()
```

## 8. Survival by passenger class

```python
class_survival = df.groupby("pclass")["survived"].mean() * 100
print("Survival rate by passenger class (%):")
display(class_survival.round(2).to_frame("Survival rate (%)"))

plt.figure(figsize=(7, 5))
sns.barplot(data=df, x="pclass", y="survived", errorbar=None)
plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger class (1 = first, 2 = second, 3 = third)")
plt.ylabel("Survival rate")
plt.ylim(0, 1)
plt.show()
```

## 9. Passenger age distribution

```python
plt.figure(figsize=(9, 5))
sns.histplot(data=df, x="age", bins=30, kde=True)
plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Number of passengers")
plt.show()

print("Average age:", round(df["age"].mean(), 2))
print("Median age:", round(df["age"].median(), 2))
```

## 10. Age compared with survival

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="survived", y="age")
plt.title("Passenger Age vs Survival")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Age")
plt.show()
```

## 11. Ticket fare analysis

```python
plt.figure(figsize=(9, 5))
sns.histplot(data=df, x="fare", bins=30, kde=True)
plt.title("Ticket Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Number of passengers")
plt.show()

print("Average fare:", round(df["fare"].mean(), 2))
print("Median fare:", round(df["fare"].median(), 2))
print("Minimum fare:", round(df["fare"].min(), 2))
print("Maximum fare:", round(df["fare"].max(), 2))
```

## 12. Correlation heatmap

```python
numeric_df = df.select_dtypes(include=np.number)
correlation = numeric_df.corr()

plt.figure(figsize=(11, 7))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap of Numeric Features")
plt.show()
```

## 13. Survival by embarkation port

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="embarked", hue="survived")
plt.title("Survival Count by Embarkation Port")
plt.xlabel("Embarkation port (C = Cherbourg, Q = Queenstown, S = Southampton)")
plt.ylabel("Number of passengers")
plt.legend(title="Survived", labels=["No", "Yes"])
plt.show()

port_survival = df.groupby("embarked", observed=False)["survived"].mean() * 100
display(port_survival.round(2).to_frame("Survival rate (%)"))
```

## 14. Family size analysis

```python
df["family_size"] = df["sibsp"] + df["parch"] + 1

family_survival = df.groupby("family_size")["survived"].mean() * 100
display(family_survival.round(2).to_frame("Survival rate (%)"))

plt.figure(figsize=(9, 5))
sns.barplot(x=family_survival.index, y=family_survival.values, errorbar=None)
plt.title("Survival Rate by Family Size")
plt.xlabel("Family size (including passenger)")
plt.ylabel("Survival rate (%)")
plt.show()
```

## 15. Final statistical summary

```python
print("========== TITANIC EDA REPORT ==========")
print("Total passenger records:", len(df))
print("Survived:", int((df["survived"] == 1).sum()))
print("Did not survive:", int((df["survived"] == 0).sum()))
print("Overall survival rate:", round(df["survived"].mean() * 100, 2), "%")
print("Average age:", round(df["age"].mean(), 2))
print("Average ticket fare:", round(df["fare"].mean(), 2))

print("\nSurvival rate by gender (%):")
print((df.groupby("sex", observed=False)["survived"].mean() * 100).round(2))

print("\nSurvival rate by passenger class (%):")
print((df.groupby("pclass")["survived"].mean() * 100).round(2))
print("=======================================")
```
